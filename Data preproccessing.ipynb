```python
#Importing Liberaries
import pandas as pd
import numpy as np
```

```python
#1-Removing Duplicates

data={'Names': ['Mohammed','Ali','Mohammed','Ali'] , 'Age':[19,18,19,19],
      'Gender':['male','female','male','female']}
df=pd.DataFrame(data)
print('Initial Data:')
display(df)
print("============="*2)
df_cleaned=df.drop_duplicates()
print("Data after Cleaning:")
display(df_cleaned)

#using keep parameter
df_cleaned=df.drop_duplicates(keep='last',)
print("data keeping last")
display(df_cleaned)

#Inplace parameter
print("the usage of inplace parameter")
display(df.drop_duplicates( inplace=False))
display(df)

#subest parameter
print ("the usage of subest parameter")
df_cleaned=df.drop_duplicates(subset='Age')
display(df_cleaned)
```

```python
#2-Handeling Missing Values
#filling
Data={'Name':['Mohammed',None,'Walter White','Raiden']
     ,'Age':[19,18,42,None]
     ,'Place':['Menof','Menof','Albequrqy',None]}
df=pd.DataFrame(Data)
display(df)
df['Name'].fillna('Unknown',inplace=True)
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Place'].fillna('Undefined',inplace=True)
df
```

```python
# 2 - drop Missing Values
data = {
    'Name': ['Ali', 'Sara', 'Omar', 'Nada', np.nan],
    'Age': [25, 30, np.nan, 22, 28],
    'Gender': ['M', 'F', 'M', np.nan, 'F'],
    'Salary': [5000, 6000, 5500, 5200, np.nan],
    'Department': ['HR', 'IT', 'HR', 'IT', 'Finance']
}
df=pd.DataFrame(data)
display(df)
df.dropna(inplace=False) #keep Original
display(df)
df.dropna(inplace=True) #Delete Original
display(df)
```

```python
#3-Correct DataTypes
data = {'Date': ['2023-01-01', '2023-02-01', '2023-03-01'], 'Sales': [100, 150, 200]} 
df=pd.DataFrame(data)
display(df)
display(df.dtypes)
df['Date'] = pd.to_datetime(df['Date']) 
display(df.dtypes)
```

```python
#4-Standardize Data
data = {'Name': ['alice', 'BOB', 'Charlie'], 'Age': [25, 30, 35]}
df=pd.DataFrame(data)
display(df)
df['Name']=df['Name'].str.title()
display(df)
```

```python
#5-Filter Outliers 
data = {'Values': [10, 12, 1000, 13]}
df=pd.DataFrame(data)
print("\033[31mOriginal DataFrame\033[0m")
display(df) 
Q1=df['Values'].quantile(0.25)
Q3=df['Values'].quantile(0.75)
IQR=Q3-Q1
lower_whisker=Q1-1.5*IQR
upper_whisker=Q3+1.5*IQR
df_cleaned=df[(df['Values']>=lower_whisker) & (df['Values']<=upper_whisker)]
display(df_cleaned)
```

```python
#6-Normalize Data
data = {'Feature1': [100, 200, 300, 500], 'Feature2': [0.5, 0.2, 0.3,.9]}
df=pd.DataFrame(data)
display(df)
df_Normalized=(df-df.min())/(df.max()-df.min())
display(df_Normalized)
```

```python
data = {
    'Name': ['Ali', 'Sara', 'Omar', 'Nada', np.nan],
    'Age': [25, 30, np.nan, 22, 28],
    'Gender': ['M', 'F', 'M', np.nan, 'F'],
    'Salary': [5000, 6000, 5500, 5200, np.nan],
    'Department': ['HR', 'IT', 'HR', 'IT', 'Finance']}
df=pd.DataFrame(data)
display(df)
df_normalized=df
df_normalized['Age']=(df['Age']-df['Age'].min())/(df['Age'].max()-df['Age'].min())
df_normalized['Salary']=(df['Salary']-df['Salary'].min())/(df['Salary'].max()-df['Salary'].min())
display(df_normalized)
```

```python
#7. Remove Irrelevant Data
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35], 'Unnecessary': [1, 2, 3]}
df=pd.DataFrame(data)
display(df)
df_cleaned=df.drop(columns=['Unnecessary'])
display(df_cleaned)
```

```python
#8. Validate Data Integrity
data = {'Name': ['Alice', 'Bob', 'Charly'], 'Age': [25, 30, 35]}
valid_names = ['Alice', 'Bob', 'Charlie']
df = pd.DataFrame(data)
display(df)
df_cleaned=df[df['Name'].isin(valid_names)]
display(df_cleaned)
```

```python
#9. Document Cleaning Steps
data = {'Name': ['Alice', 'Bob', 'Charlie', 'Alice'], 'Age': [25, 30, 35, np.nan], 'Salary': [50000, 60000, 75000, 50000]} 
df=pd.DataFrame(data)
display(df)
cleaning_steps=[]
df_cleaned=df.drop_duplicates(subset=['Name'])
cleaning_steps.append('Removed Duplicate rows')
display(df_cleaned)
df_cleaned['Age']=df_cleaned['Age'].astype(int)
df_cleaned['Age']=df_cleaned['Age'].astype(int)
cleaning_steps.append('datatypes corrected')
display(df_cleaned.dtypes)
```

```python
#11.1.Label encoding
data = {
    'Name': ['Ali', 'Sara', 'Omar', 'Nada', np.nan],
    'Age': [25, 30, np.nan, 22, 28],
    'Gender': ['M', 'F', 'M', np.nan, 'F'],
    'Salary': [5000, 6000, 5500, 5200, np.nan],
    'Department': ['HR', 'IT', 'HR', 'IT', 'Finance']
}
df=pd.DataFrame(data)
df['Gender'].fillna(value='unknown',inplace=True)
df['Gender_encoded']=df['Gender'].map({'M':1,'F':0,'unknown':-1})
df
```

```python
#label encoding using sklearn
df=df.drop(columns='Gender_encoded')
display(df)
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Gender_encoded'] = encoder.fit_transform(df['Gender'])
print(df)
```

```python
df=df.drop(columns='Gender_encoded')
display(df)
df=pd.get_dummies(df,columns=['Gender'])
display(df)
```
